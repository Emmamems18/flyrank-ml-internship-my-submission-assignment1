## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Unit of Analysis: A single piece of content (content_id).

Decision Improved: Deciding which specific "striking distance" pages (currently ranking on Page 2 or 3) should be prioritized for immediate SEO optimization resources.

Who Acts & Action Taken: The SEO Specialist or Content Manager. They will take direct action by updating on-page keywords, refreshing content depth, and building internal links for the flagged pages.

Cost of a Wrong Call:
False Positive (Recommending a low-potential page): Wastes valuable editorial hours and marketing budget on articles that fail to move up to Page 1.

False Negative (Missing a high-potential page): Leaves easy, high-intent organic traffic on the table, allowing competitors to capture Page 1 market share instead.

```python
# This cell is for CODE (numbers, a query, a check).
# Check total content pieces vs striking distance opportunities (Page 2 & 3 rankings)
total_pages = len(df)
striking_distance_pages = df[
    df["position_tier"].isin(["page_2_3", "page_3_5"])
    | ((df["avg_position"] >= 11) & (df["avg_position"] <= 30))
]

count_striking = len(striking_distance_pages)
pct_striking = (count_striking / total_pages) * 100
avg_search_vol = striking_distance_pages["search_volume"].mean()

print(f"Total Pages in Dataset: {total_pages:,}")
print(
    f"Striking Distance Pages (Positions 11-30): {count_striking:,}"
    f" ({pct_striking:.1f}%)"
)
print(
    f"Average Monthly Search Volume for Striking Pages:"
    f" {avg_search_vol:.0f} searches"
)
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

What this work CAN claim:

Observed patterns: Identifies empirical relationships in historical data showing which page attributes co-occur with better ranking performance.

Directional guidance: Provides relative prioritization scores to rank the 10,201 striking-distance candidates from highest to lowest expected return on effort.

Decision-support: Reduces human effort by narrowing down thousands of unorganized pages into an actionable shortlist for editorial teams.

What this work CANNOT claim:

Causal proof: Observed correlations do not prove causality. Increasing a page's word count or updating its title does not guarantee Google will move it to Page 1.

Predicting Google's algorithm: Google uses hundreds of proprietary, real-time, and external signals (e.g., backlinks, user search history, domain authority) not present in this dataset.

Guaranteed outcomes: Models support decision-making but cannot guarantee specific position jumps or exact traffic gains.

```python
# This cell is for CODE (numbers, a query, a check).
# Check correlations with average position to illustrate directional trends vs. noise
correlation_cols = [
    "avg_position",
    "word_count",
    "ctr",
    "days_since_last_update",
    "engagement_rate",
]
correlations = df[correlation_cols].corr()["avg_position"].drop("avg_position")

print("Correlation with Average Position:")
print(correlations.round(3))
print(
    "\nNote: Low to moderate correlation values show directional signals,"
    " not deterministic rules."
)
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```
